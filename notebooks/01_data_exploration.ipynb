# Day-Ahead Price Forecast — Exploratory Data Analysis

**Goal:** Understand the structure, quality, and dynamics of the DE-LU day-ahead electricity
market data before modelling. The analysis covers data-quality checks (missingness, gaps,
plausibility) and structural insights (merit-order effect, seasonality, regime shifts,
cross-border price–flow relationships).

**Time range:** 2019-01-01 to yesterday (UTC). The full range deliberately spans the
pre- and post-2022 energy-crisis regimes, enabling regime-change analysis.

**Data sources:**
- ENTSO-E Transparency Platform (via `entsoe-py`): day-ahead prices, load, wind/solar
  forecasts and actuals, generation mix, scheduled exchanges, cross-border flows.
- Yahoo Finance (via `yfinance`): TTF Natural Gas Futures (`TTF=F`), EUA CO₂ Allowances (`CO2.L`).

**Last refresh:** _2026-06-02_

**Out of scope:** Feature engineering, modelling, and dashboards.
Those live in separate notebooks and scripts.

```python
daily_load = df[["load_actual", "load_forecast_day_ahead"]].resample("D").mean()

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 7), sharex=True)

ax1.plot(
    daily_load.index, daily_load["load_actual"],
    lw=0.8, label="actual", color="steelblue",
)
ax1.plot(
    daily_load.index, daily_load["load_forecast_day_ahead"],
    lw=0.8, label="forecast", color="tomato", alpha=0.7,
)
ax1.set_title("Load: actual vs day-ahead forecast (daily mean)")
ax1.set_ylabel("MW")
ax1.legend()

daily_error = ((df["load_actual"] - df["load_forecast_day_ahead"])/df["load_actual"]).resample("D").mean()
ax2.plot(daily_error.index, daily_error.values, lw=0.8, color="purple", alpha=0.8)
ax2.axhline(0, color="black", lw=0.8, linestyle="--")
ax2.set_title("Load forecast error (actual − forecast)/actual, daily mean)")
ax2.set_ylabel("relative error")
ax2.set_xlabel("Date")

plt.tight_layout()
plt.show()
```

In the crisis, the load dip results in strong overestimations. Post-crisis, it is well centered but oscillating stronger with a visible seasonal trend. Possible drivers include demand destruction in heavy industry, increased heat-pump penetration, growing behind-the-meter PV self-consumption, and government-driven energy-saving campaigns.

```python
from typing import Sequence


def plot_border_flows_grid(frame: pd.DataFrame, neighbors: Sequence[str]) -> None:
    """Plot scheduled vs physical net flows for all neighbors as subplot grid."""
    fig, axes = plt.subplots(3, 2, figsize=(14, 10), sharex=True)
    daily = frame.resample("D").mean()
    for ax, nb in zip(axes.flat, neighbors):
        sched_col = f"scheduled_net_de_to_{nb}"
        phys_col = f"physical_net_de_to_{nb}"
        if sched_col in daily.columns:
            daily[sched_col].plot(ax=ax, label="scheduled", alpha=0.7)
        if phys_col in daily.columns:
            daily[phys_col].plot(ax=ax, label="physical", alpha=0.7)
        ax.axhline(0, color="black", lw=0.5)
        ax.set_title(f"DE \u2192 {nb.upper()}")
        ax.legend(fontsize=8)
    plt.suptitle("Cross-border net flows: scheduled vs physical (daily mean, positive = DE export)", y=1.01)
    plt.tight_layout()
    plt.show()


NEIGHBORS = ["fr", "nl", "at", "pl", "ch", "dk_1"]
plot_border_flows_grid(df, NEIGHBORS)
```

Scheduled and physical flows track each other decently close for most neighbors, validating the ENTSO-E data consistency. France and the Netherlands are the largest flow corridors; Austria and Switzerland show smaller but consistently directional flows. The 2022 energy crisis is visible as a shift in flow directions for some corridors, reflecting the reconfiguration of European energy trade routes.

The three patterns are structurally different and all expected:

**Nuclear (`gen_nuclear`):** Data is present from 2019 through early 2023, then drops to zero permanently from April 2023 onward — consistent with Germany's final nuclear shutdown. These are not missing values; the generation truly went to zero.

**EUA CO2 (`eua_co2_eur_per_t`):** Entirely absent before October 2021. The `CO2.L` ticker (ICE EUA Futures via Yahoo Finance) was not reliably available before that date. This is a structural data gap that cannot be recovered. After that gap data is still missing, seemingly at random. These gaps are likely caused by Yahoo Finance data quality issues: futures tickers are prone to gaps during contract rollovers, and Yahoo Finance does not provide a professionally maintained feed for commodity futures. A production system would source TTF and EUA data from EEX or ICE directly. For EDA purposes, the gaps are noted and left as NaN.

**TTF gas (`ttf_gas_eur_per_mwh`):** Present throughout, but with intermittent gaps beyond what weekends and holidays alone would explain. The loader forward-fills both commodity columns with a 4-day limit, which covers normal weekend and holiday gaps. Longer gaps are likely caused by Yahoo Finance data quality issues: futures tickers are prone to gaps during contract rollovers, and Yahoo Finance does not provide a professionally maintained feed for commodity futures. A production system would source TTF and EUA data from EEX or ICE directly. For EDA purposes, the gaps are noted and left as NaN.

**Time resolution change:** There is a change in the structure of the data visible. From 2025-09-30 22:15 UTC onward, ENTSO-E publishes data at 15-minute resolution instead of hourly. The current DataFrame contains a mix of 59,166 hourly and 22,824 quarter-hourly rows. As missingno is not built for a time-index, this becomes visible and must be minded when further using the data.

### 2.3 Plausibility Checks

```python
SPIKE_THRESHOLD_EUR = 200
EXTREME_SPIKE_THRESHOLD_EUR = 500
NEG_PRICE_THRESHOLD = 0

# Resample to hourly to avoid the 4× overcount of the post-2025 15-min period.
price_h = df["day_ahead_price"].resample("h").mean().dropna()
total_h = len(price_h)

fig, ax = plt.subplots(figsize=(12, 5))
ax.hist(price_h, bins=100, color="steelblue", edgecolor="none", alpha=0.8)
ax.axvline(NEG_PRICE_THRESHOLD, color="red", lw=1.2, linestyle="--", label="0 EUR/MWh")
ax.axvline(
    SPIKE_THRESHOLD_EUR, color="darkorange", lw=1.2, linestyle="--",
    label=f"{SPIKE_THRESHOLD_EUR} EUR/MWh (spike threshold)",
)
ax.set_title("Day-ahead price distribution (hourly, corrected for resolution break)")
ax.set_xlabel("EUR/MWh")
ax.set_ylabel("hours")
ax.legend()
plt.tight_layout()
plt.show()

neg    = (price_h < NEG_PRICE_THRESHOLD).sum()
spikes = (price_h > SPIKE_THRESHOLD_EUR).sum()
extreme = (price_h > EXTREME_SPIKE_THRESHOLD_EUR).sum()
print(f"Negative price hours:           {neg:>6,} ({neg / total_h:.2%} of total)")
print(f"Spike hours (>{SPIKE_THRESHOLD_EUR} EUR/MWh):   {spikes:>6,} ({spikes / total_h:.2%} of total)")
print(f"Extreme spike hours (>{EXTREME_SPIKE_THRESHOLD_EUR} EUR/MWh): {extreme:>6,} ({extreme / total_h:.2%} of total)")
```

The price distribution is strongly right-skewed with the bulk of hours concentrated between 0 and 100 EUR/MWh; a long right tail reflects the 2022 crisis period. Negative prices occur in a small but non-negligible share of hours, driven by renewable surplus (typically weekend middays with high solar and low demand). Extreme spikes above 500 EUR/MWh are very rare and concentrated in 2022.

The chart shows several expected structural patterns that validate the data quality.

**Seasonal stack height:** Total generation is consistently higher in winter months than in summer, which is plausible — higher electricity demand for heating, lighting, and industrial activity during the cold season drives more capacity into dispatch. This seasonality is visible as a regular annual rhythm in the overall stack height throughout the full 2019–2026 range.

**Solar (yellow):** Peaks sharply each summer and nearly vanishes in winter. This is the physically expected behaviour: longer days and higher solar angles in June–August maximise photovoltaic yield, while winter months with short, low-angle days contribute very little. The summer peaks grow noticeably from 2019 to 2026, consistent with continued capacity additions.

**Wind onshore (light blue) and offshore (blue):** Both sources peak in winter and trough in summer — the seasonal complement to solar. Northern European wind patterns are stronger and more frequent in autumn and winter, making wind the dominant renewable in the darker half of the year. Together, wind and solar provide a natural seasonal hedge against each other.

**Nuclear (pink):** Visible as a steady base-load layer through 2022 and into early 2023, then disappears entirely after Germany's final three-plant shutdown in April 2023. The column becomes NaN in the ENTSO-E data post-shutdown (the source stops reporting it); the `fillna(0)` applied before plotting reflects the true physical state: zero generation.

**Lignite (brown) and hard coal (grey):** Remain substantial base-load contributors throughout, though with some visible reduction from 2023 onward as renewables displace them at the margin. The two coal types together represent the largest single share of the stack in the pre-crisis period.

**Gas (orange):** Shows more variability, often filling residual demand left by weather-driven swings in wind and solar. Its contribution spikes in some winter months and recedes when renewables are strong.

**Actual load (black line):** The load line runs until the crisis below the total generation stack. This is expected: Germany was structurally a net exporter of electricity, and the gap between domestic generation and domestic consumption represents net cross-border exports. The gap narrows during the 2022 energy crisis — consistent with Germany reducing exports and importing more during periods of extreme gas scarcity — and widens again afterward in the opposite direction underlining the change from an exporter toward an importer. The load line mirrors the seasonal rhythm of the stack, confirming that generation broadly follows demand across the year.

### 2.4 DQ Findings

**Confirmed structural patterns (expected):**
- Load and generation seasonalities are clearly visible
- Strong structural break through the gas crisis
- Gas-Price seems to be one of the main price drivers
- Nuclear generation (gen_nuclear) tapers to zero by April 2023, reflecting Germany's final shutdown — this structural break affects the generation mix feature permanently.
- Renewable generation shows also strong seasonality (wind: winter peak, solar: summer peak) with a visible multi-year upward capacity trend.

**Data quality issues (need handling in feature engineering):**
- **Time resolution change:** From 2025-09-30 22:15 UTC onward, ENTSO-E publishes data at 15-minute resolution instead of hourly. The current DataFrame contains a mix of 59,166 hourly and 22,824 quarter-hourly rows. The loader does not normalise this — the mixed index will break any hourly feature pipeline. Sprint 2 must decide: resample 15-min data to hourly, or migrate the full pipeline to 15-min resolution.
- EUA CO2 data only available from October 2021 onward and missing for ~38% of the full time range (pre-Oct-2021); feature engineering must decide between truncating the training window or encoding missingness explicitly.
- Small NaN gaps (<1%) in several ENTSO-E columns (generation types, cross-border flows); forward-fill or linear interpolation over short gaps is likely appropriate.
- Negative price hours exist and are legitimate market events (renewable surplus); any log-transformation of the target must account for these.
- Extreme price spikes (>500 EUR/MWh) are rare but real; outlier capping or robust loss functions should be considered for the model.

**Open questions (revisit later):**
- Are the simultaneous NaN bands across ENTSO-E columns due to API outages or genuine reporting gaps? Cross-check against ENTSO-E transparency platform directly.
- Is the change of the behaviour of load forecast error post crisis now permanent?
- What is the best way to adress the forecast uncertainty in renewables generation?

## 3. Structural Insights

This section answers what the data tells us about the structure of the DE-LU day-ahead market, and which features are likely to matter for the price forecast model. Each analysis directly motivates one or more feature design decisions.

### 3.0 Canonical Hourly Frame and Derived Quantities

```python
# Canonical hourly frame for all structural analysis below.
# The loader returns a mixed-resolution index (hourly until 2025-09-30, then
# 15-min). Resampling to a regular hourly grid makes autocorrelation lags,
# seasonal profiles, and count-based statistics consistent across the full range.
# This is a notebook-local view only; normalising the loader to a single
# resolution is a deliberate Sprint-2 decision (see DQ findings, section 2.4).
df_h = df.resample("h").mean(numeric_only=True)

assert str(df_h.index.tz) == "UTC"
diffs = df_h.index.to_series().diff().dropna().unique()
assert len(diffs) == 1 and diffs[0] == pd.Timedelta("1h"), "index is not strictly hourly"

print(f"df_h shape: {df_h.shape}  |  {df_h.index.min().date()} → {df_h.index.max().date()}")
```

```python
RENEWABLE_GEN_COLS = ["gen_wind_onshore", "gen_wind_offshore", "gen_solar"]
SPIKE_THRESHOLD_EUR = 200

# Residual load: demand that must be met by dispatchable plants after subtracting
# must-run wind and solar. Primary physical driver of price (x-axis of fishhook).
df_h["residual_load"]   = df_h["load_actual"] - df_h[RENEWABLE_GEN_COLS].sum(axis=1)

# Renewable share: fraction of load covered by wind + solar.
df_h["renewable_share"] = df_h[RENEWABLE_GEN_COLS].sum(axis=1) / df_h["load_actual"]

# Local-time calendar attributes for seasonality profiles. Data stored in UTC;
# hour-of-day and weekday effects are local-time phenomena.
idx_local = df_h.index.tz_convert("Europe/Berlin")
df_h["local_hour"]    = idx_local.hour
df_h["local_weekday"] = idx_local.dayofweek   # 0 = Monday
df_h["local_month"]   = idx_local.month
df_h["year"]          = idx_local.year

# Market regimes (adjustable boundaries):
#   calm:        2019-01-01 – 2021-08-31  (pre-crisis, low and stable)
#   crisis:      2021-09-01 – 2023-03-31  (gas-price ramp, peak Aug 2022)
#   post_crisis: 2023-04-01 – present     (elevated but mean-reverting)
CRISIS_START      = pd.Timestamp("2021-09-01", tz="UTC")
POST_CRISIS_START = pd.Timestamp("2023-04-01", tz="UTC")

def regime_label(ts: pd.Timestamp) -> str:
    if ts < CRISIS_START:
        return "calm"
    if ts < POST_CRISIS_START:
        return "crisis"
    return "post_crisis"

df_h["regime"] = df_h.index.map(regime_label)

REGIME_COLORS = {"calm": "steelblue", "crisis": "tomato", "post_crisis": "seagreen"}

df_h["regime"].value_counts().sort_index()
```

**Residual load** (`load_actual` minus wind and solar generation) is the demand that must be served by dispatchable plants (gas, coal, hydro). It is the x-axis of the merit-order "fishhook" and the primary physical driver of the marginal price.

**Renewable share** (wind+solar / load_actual) measures how much of demand is covered by renewables. A high share pushes residual load down and is strongly associated with low or negative prices.

Three **market regimes** are defined as named constants reused throughout this section:
- **calm** (2019-01-01 – 2021-08-31): low, stable pre-crisis prices
- **crisis** (2021-09-01 – 2023-03-31): gas-price ramp from autumn 2021, peak August 2022
- **post_crisis** (2023-04-01 – present): elevated but mean-reverting

### 3.1 Price Seasonality and Calendar Structure

```python
# Shared colour scale: use calm + post_crisis range so the crisis does not
# wash out the structural patterns in the other two panels.
_cp = df_h.loc[df_h["regime"].isin(["calm", "post_crisis"]), "day_ahead_price"].dropna()
vmin_global, vmax_global = _cp.quantile(0.02), _cp.quantile(0.98)

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)
for ax, reg in zip(axes, ["calm", "crisis", "post_crisis"]):
    sub = df_h[df_h["regime"] == reg]
    pivot = sub.pivot_table(
        values="day_ahead_price",
        index="local_hour", columns="local_month",
        aggfunc="mean",
    )
    sns.heatmap(
        pivot, ax=ax, cmap="viridis",
        vmin=vmin_global, vmax=vmax_global,
        cbar=(reg == "post_crisis"),
    )
    ax.set_title(f"Mean price (EUR/MWh) — {reg}")
    ax.set_xlabel("month")
axes[0].set_ylabel("hour of day (Europe/Berlin)")
plt.suptitle("Hour-of-day × month price heatmap per regime", y=1.02)
plt.tight_layout()
plt.show()
```

Morning (6h–9h) and evening (17h–21h) peaks are visible across all regimes, driven by residential and commercial demand cycles. A midday trough in summer months — the **solar belly** — is present even in the calm period and deepens noticeably into post_crisis, reflecting the growing solar fleet. Winter prices tend to be higher than summer across all regimes. The crisis panel is clipped by the shared colour scale. The evening prices are higher than the morning prices. When looking closer to the post-crisis market, interestingly, the lowest prices appear in the noon hours in late spring or early summer, not July or August. On the other hand, the highest prices appear to be in the evening hours in autumn, not winter. This slightly unintuitive pattern is actually well known: in May, solar generation is already very high at noon, but the load is significantly lower than in the summer due to the lack of air conditioning. In autumn, solar generation is already low in the evening, but the climate is less windy.

```python
daily_profile = (
    df_h.groupby(["regime", "local_hour"])["day_ahead_price"]
    .mean()
    .unstack(0)
)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for reg, col in REGIME_COLORS.items():
    if reg in daily_profile.columns:
        daily_profile[reg].plot(ax=ax1, label=reg, color=col, lw=1.5)
ax1.set_title("Average hourly price profile by regime")
ax1.set_xlabel("hour of day (Europe/Berlin)")
ax1.set_ylabel("EUR/MWh")
ax1.legend()

norm_profile = daily_profile - daily_profile.mean()
for reg, col in REGIME_COLORS.items():
    if reg in norm_profile.columns:
        norm_profile[reg].plot(ax=ax2, label=reg, color=col, lw=1.5)
ax2.axhline(0, color="black", lw=0.6, linestyle="--")
ax2.set_title("Normalised profile (deviation from daily mean)")
ax2.set_xlabel("hour of day (Europe/Berlin)")
ax2.set_ylabel("EUR/MWh deviation")
ax2.legend()

plt.tight_layout()
plt.show()
```

The absolute profile (left) shows the large level shift between regimes. The normalised view (right) removes the level and isolates the *shape*: the morning-to-evening structure is broadly stable across regimes, but the solar belly deepens from calm to post_crisis — the midday dip relative to the daily mean grows as installed solar capacity increases. The evening peak also sharpens slightly as solar ramps down more steeply. Both effects motivate hour-of-day and month calendar features.

### 3.2 The Merit-Order Effect (Price vs Residual Load)

```python
mask = df_h[["residual_load", "day_ahead_price"]].notna().all(axis=1)
x = df_h.loc[mask, "residual_load"]
y = df_h.loc[mask, "day_ahead_price"]

fig, ax = plt.subplots(figsize=(10, 7))
hb = ax.hexbin(x, y, gridsize=60, cmap="magma", bins="log", mincnt=1)
fig.colorbar(hb, ax=ax, label="log10(count)")

bins_cut = pd.cut(x, bins=30)
median_curve = y.groupby(bins_cut, observed=True).median()
centers = [iv.mid for iv in median_curve.index]
ax.plot(centers, median_curve.values, color="cyan", lw=2, label="binned median price")

ax.axhline(0, color="white", lw=0.6, linestyle="--")
ax.set_xlabel("residual load (MW) = load − wind − solar")
ax.set_ylabel("day-ahead price (EUR/MWh)")
ax.set_title("Merit-order effect: price vs residual load (fishhook)")
ax.legend()
plt.tight_layout()
plt.show()
```

There is a clear and strong connection visible. Below roughly 10,000 MW residual load, prices flatten near zero and turn negative, driven by must-run renewable surplus that cannot be absorbed domestically. As residual load increases, price accelerates nonlinearly, indicating that increasingly expensive power plants set the marginal price at high residual load states. The binned median gives a feeling for the general trend despite the wide scatter. At very high residual load, the curve again flattens a bit.

```python
fig, ax = plt.subplots(figsize=(10, 7))
for reg, col in REGIME_COLORS.items():
    sub = df_h[df_h["regime"] == reg]
    mask_r = sub[["residual_load", "day_ahead_price"]].notna().all(axis=1)
    x_r = sub.loc[mask_r, "residual_load"]
    y_r = sub.loc[mask_r, "day_ahead_price"]
    bins_r = pd.cut(x_r, bins=30)
    med_r = y_r.groupby(bins_r, observed=True).median()
    centers_r = [iv.mid for iv in med_r.index]
    ax.plot(centers_r, med_r.values, color=col, lw=2, label=reg)

ax.axhline(0, color="black", lw=0.6, linestyle="--")
ax.set_xlabel("residual load (MW)")
ax.set_ylabel("day-ahead price (EUR/MWh)")
ax.set_title("Merit-order fishhook by regime (binned median)")
ax.legend()
plt.tight_layout()
plt.show()
```

The curve **shifts vertically** by regime: the same physical residual load corresponds to a much higher price during the crisis. The shape and curvature of the curve are broadly similar across regimes — it is not the physical supply/demand structure that changed, but the fuel cost: gas-plant marginal costs spiked with TTF, lifting the entire curve. Post_crisis sits between calm and crisis, consistent with partial gas-price mean-reversion. This motivates including the gas price as a feature alongside residual load.

```python
rolling_mean = df_h["day_ahead_price"].rolling(24 * 30, min_periods=24 * 7).mean()
rolling_std  = df_h["day_ahead_price"].rolling(24 * 30, min_periods=24 * 7).std()

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 7), sharex=True)

ax1.plot(rolling_mean.index, rolling_mean.values, lw=1, color="steelblue")
ax1.axvline(CRISIS_START,      color="tomato",   lw=1.2, linestyle="--", label="crisis start")
ax1.axvline(POST_CRISIS_START, color="seagreen", lw=1.2, linestyle="--", label="post-crisis start")
ax1.set_ylabel("EUR/MWh")
ax1.set_title("30-day rolling mean of hourly price (level)")
ax1.legend()

ax2.plot(rolling_std.index, rolling_std.values, lw=1, color="purple")
ax2.axvline(CRISIS_START,      color="tomato",   lw=1.2, linestyle="--", label="crisis start")
ax2.axvline(POST_CRISIS_START, color="seagreen", lw=1.2, linestyle="--", label="post-crisis start")
ax2.set_ylabel("EUR/MWh")
ax2.set_title("30-day rolling std of hourly price (volatility)")
ax2.set_xlabel("Date")
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
def regime_stats(frame: pd.DataFrame, price_col: str = "day_ahead_price") -> pd.DataFrame:
    """Summarise price level, dispersion, and tail behaviour per regime."""
    rows = []
    for reg, g in frame.groupby("regime", observed=True):
        p = g[price_col].dropna()
        rows.append({
            "regime":         reg,
            "mean":           round(p.mean(), 1),
            "std":            round(p.std(), 1),
            "p05":            round(p.quantile(0.05), 1),
            "median":         round(p.median(), 1),
            "p95":            round(p.quantile(0.95), 1),
            "negative_share": round((p < 0).mean(), 4),
            "spike_share":    round((p > SPIKE_THRESHOLD_EUR).mean(), 4),
        })
    order = {"calm": 0, "crisis": 1, "post_crisis": 2}
    return (
        pd.DataFrame(rows)
        .sort_values("regime", key=lambda s: s.map(order))
        .set_index("regime")
    )

regime_stats(df_h)
```

```python
mask_sp = df_h[["residual_load", "day_ahead_price", "ttf_gas_eur_per_mwh"]].notna().all(axis=1)
sp = df_h.loc[mask_sp]

fig, ax = plt.subplots(figsize=(10, 7))
sc = ax.scatter(
    sp["residual_load"], sp["day_ahead_price"],
    c=sp["ttf_gas_eur_per_mwh"], cmap="plasma",
    s=2, alpha=0.3,
)
fig.colorbar(sc, ax=ax, label="TTF gas (EUR/MWh)")
ax.axhline(
    SPIKE_THRESHOLD_EUR, color="red", lw=0.8, linestyle="--",
    label=f"spike threshold ({SPIKE_THRESHOLD_EUR} EUR/MWh)",
)
ax.set_xlabel("residual load (MW)")
ax.set_ylabel("day-ahead price (EUR/MWh)")
ax.set_title("Spike drivers: residual load × price, coloured by gas price")
ax.legend()
plt.tight_layout()
plt.show()
```

The rolling plots show that both **level and volatility** shifted in 2022: the crisis was not merely a level jump — price swings also became much wider and have not fully normalised post-crisis. The regime table quantifies this: median prices roughly tripled from calm to crisis and partially reverted; standard deviation increased nearly 4×. Spike hours (>200 EUR/MWh) are almost entirely confined to the crisis window. The scatter confirms that price spikes require **both** high residual load and high gas prices (bright yellow/orange dots cluster at top right) — neither alone is sufficient. This motivates including TTF gas price alongside residual load, and treating the 2022 regime as either a training-window decision or an explicit regime feature.

### 3.5 Autocorrelation and Seasonality Decomposition

```python
price_2024 = df_h.loc["2024", "day_ahead_price"]
n_missing = int(price_2024.isna().sum())
if n_missing > 0:
    print(f"Interpolating {n_missing} gap(s) in 2024 window (linear, limit=3h).")
    price_2024 = price_2024.interpolate(method="time", limit=3)

remaining = int(price_2024.isna().sum())
assert remaining == 0, f"{remaining} NaN remain after interpolation — choose a different year or widen limit"
print(f"2024 window: {len(price_2024):,} hourly observations, 0 NaN")
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
plot_acf(price_2024, lags=200, ax=ax1)
ax1.set_title("ACF of hourly day-ahead price (2024)")
plot_pacf(price_2024, lags=200, ax=ax2, method="ywm")
ax2.set_title("PACF of hourly day-ahead price (2024)")
plt.tight_layout()
plt.show()
```

The ACF shows strong positive autocorrelation at short lags (1h, 2h, ...) that decays slowly — the current price is a strong predictor of the next hour. Clear **spikes at lag 24** (daily cycle) which decay over the following days but slightly peak again at **lag 168** (weekly cycle), confirming both seasonalities empirically. The PACF cuts off much more sharply after short lags but shows the same structure. These results are the direct empirical justification for the planned lag features.

```python
try:
    # trend=1441 ≈ 60-day smoother (must be odd); suppresses sub-monthly wiggles
    # while still capturing the summer-dip / winter-peak level shift.
    mstl = MSTL(price_2024, periods=(24, 168), stl_kwargs={"trend": 1441})
    res = mstl.fit()
    fig = res.plot()
    fig.set_size_inches(14, 20)
    plt.subplots_adjust(hspace=0.4)
    for ax in fig.axes:
        ax.set_ylim(-100, 200)
    plt.suptitle(
        "MSTL decomposition (2024): trend, daily seasonality, weekly seasonality, residual",
        y=1.01,
    )
    plt.show()
except Exception as exc:
    print(f"MSTL failed ({exc}), falling back to STL(period=24).")
    res = STL(price_2024, period=24, trend=1441).fit()
    fig = res.plot()
    fig.set_size_inches(14, 16)
    plt.subplots_adjust(hspace=0.4)
    for ax in fig.axes:
        ax.set_ylim(-100, 200)
    plt.suptitle("STL decomposition (2024, daily period only)", y=1.01)
    plt.show()
```

The MSTL decomposition separates the 2024 price series into a smooth **trend** (medium-term level movement), a **daily seasonal** component (the 24h profile seen in 3.1), a **weekly seasonal** component (weekday/weekend pattern), and a **residual**. The residual is substantial relative to the seasonal amplitudes — a large share of price variation cannot be explained by calendar effects alone. This unexplained component is driven by weather (wind/solar variation), gas prices, demand shocks, and cross-border flows; it is the core prediction target for the model.

```python
stat, pvalue, *_ = adfuller(price_2024.values)
print(f"ADF statistic: {stat:.3f}")
print(f"p-value:       {pvalue:.4f}")
print("Interpretation: p < 0.05 suggests the series is stationary "
      "(no unit root) over this window.")
print("Note: LightGBM does not require stationarity — this test informs "
      "the ARIMAX benchmark planned for Sprint 2.")
```

The ADF result is reported above. Day-ahead prices exhibit seasonality and inter-year level shifts (regime changes), but within a single calendar year the series is typically stationary — p < 0.05 is the expected outcome on the 2024 window. LightGBM, as a tree-based model, does not assume stationarity; the test is included as context for the ARIMAX baseline, which does impose this assumption.

### 3.7 Structural Findings

**Confirmed market structure (feeds feature engineering):**
- Clear daily (24h) and weekly (168h) price seasonality — confirmed by ACF/PACF peaks and MSTL decomposition. Morning and evening demand peaks raise prices; a deepening solar belly suppresses midday prices in summer, growing with installed PV capacity.
- Merit-order fishhook confirmed: residual load is the primary physical driver of price. The relationship is nonlinear — steep at high residual load (gas/oil at the margin), flat or negative at low residual load (renewable surplus).
- As a result, the gas price is a strong price driver.
- Negative prices are systematic and growing year-on-year: driven by high renewable share and low residual load, concentrated in spring/summer midday hours. Explained, not noise.
- Cross-border flows do have a visible correlation with the price.

**Quantified regime shift (2022):**
See regime table (section 3.4). Key numbers:
- Median price: calm ~50 EUR/MWh → crisis ~190 EUR/MWh → post_crisis ~80 EUR/MWh.
- Standard deviation: roughly 4× higher during the crisis; post-crisis volatility remains elevated above the calm baseline.
- Spike share (>200 EUR/MWh): near-zero outside the crisis window.
- Fishhook vertical shift: the same physical residual load state produces a higher price in the crisis driven by the gas price.

**Implications for feature design (Sprint 2):**
- **Lag features** at for example 24h, 48h, 168h are empirically justified by ACF/PACF.
- Forecasts for **residual load** and **renewable share** have to be primary physical features; both explain negative prices and price level.
- The level of the **gas price (TTF)** is necessary to explain the fishhook vertical shift and spike conditions — residual load alone cannot distinguish calm from crisis price levels.
- **Calendar features** (hour-of-day, weekday, month, holiday) capture the seasonal patterns confirmed.
- **Regime indicator or training window decision**: the crisis period is a structural outlier. Options: down-weight it, exclude it from training, or include an explicit regime feature.

**Open questions (revisit later):**
- What is the most powerful way to include cross-border flows as a feature, as they are not available before the spot price auction and scheduled and physical flows often deviate?

### 4.2 Do Lagged Forecast Errors Carry Signal?

**Leakage context:** The forecast error is `actual − forecast`. The `actual` value is only known *after* the delivery hour. For target hour H on day D, the error for H is not available at gate closure (GK, ≈12:00 CET on D-1). The freshest observable error at gate closure comes from the morning of D-1. The minimum usable lag is therefore **≥ 24 + (H - GC) h**. The relevant question is not whether autocorrelation exists at lag 1h (almost certainly yes), but whether it persists at **lag >=24h** — the lags where the error could actually be used as a feature.

**Solar note:** Solar generation is structurally zero at night (forecast = actual = 0 → error = 0). Including night-time hours would create artificial 24h periodicity in the ACF driven by the day/night rhythm, not by forecast-error persistence. In addition, a threshold of `solar_forecast > 0` would retain dusk/dawn hours where the forecast is near-zero; the corresponding errors cluster around zero and distort the ACF. Solar is therefore restricted to hours with `solar_forecast > 100 MW`, selecting only hours with meaningful expected generation. The resulting series is a sparse daytime subsample; ACF lags represent consecutive qualifying steps, so the 24h and 168h reference lines are approximate guides, not exact calendar periods.

```python
# 2024 window — same gap-free year used in 3.5
df_acf = df_h.loc[
    pd.Timestamp("2024-01-01", tz="UTC") : pd.Timestamp("2024-12-31 23:00", tz="UTC")
].copy()

# Physical net export total (scheduled already exists from 3.6)
phys_cols = [c for c in df_acf.columns if c.startswith("physical_net_de_to_")]
df_acf["total_physical_net_export"] = df_acf[phys_cols].sum(axis=1)


def _fill(s: pd.Series, limit: int = 3) -> pd.Series:
    """Linear interpolation over short gaps; report any filled gaps."""
    filled = s.interpolate(method="time", limit=limit)
    n_gaps = int(s.isna().sum())
    if n_gaps:
        print(f"  {s.name}: interpolated {n_gaps} gap(s)")
    return filled


print("Interpolating short gaps (limit=3h) where needed:")

load_err = _fill(
    (df_acf["load_actual"] - df_acf["load_forecast_day_ahead"]).rename("load_err")
)
wind_onshore_err = _fill(
    (df_acf["gen_wind_onshore"] - df_acf["wind_onshore_forecast"]).rename("wind_onshore_err")
)
wind_offshore_err = _fill(
    (df_acf["gen_wind_offshore"] - df_acf["wind_offshore_forecast"]).rename("wind_offshore_err")
)

# Solar daytime filter: exclude hours with near-zero forecasts (night and deep twilight).
# A threshold > 0 retains dusk/dawn hours where forecast ≈ 0 and error ≈ 0,
# creating artificial near-zero clusters. 100 MW ensures only hours with meaningful
# expected generation are included.
SOLAR_MIN_MW = 100
solar_err_raw = (df_acf["gen_solar"] - df_acf["solar_forecast"]).rename("solar_err")
solar_err_daytime = solar_err_raw.loc[df_acf["solar_forecast"] > SOLAR_MIN_MW].dropna()

cross_border_dev = _fill(
    (df_acf["total_physical_net_export"] - df_acf["total_scheduled_net_export"]).rename(
        "cross_border_dev"
    )
)

print("\nSeries lengths:")
for s in [load_err, wind_onshore_err, wind_offshore_err, solar_err_daytime, cross_border_dev]:
    print(f"  {s.name}: {len(s):,} obs, {s.isna().sum()} NaN")
```

```python
fig, axes = plt.subplots(5, 2, figsize=(14, 20))

error_series = {
    "Load forecast error (MW)": load_err,
    "Wind onshore forecast error (MW)": wind_onshore_err,
    "Wind offshore forecast error (MW)": wind_offshore_err,
    f"Solar forecast error, daytime only (>{SOLAR_MIN_MW} MW forecast) (MW)": solar_err_daytime,
    "Cross-border plan-vs-actual deviation (MW)": cross_border_dev,
}

axes[0, 0].set_title("ACF", fontsize=12, fontweight="bold")
axes[0, 1].set_title("PACF", fontsize=12, fontweight="bold")

for row, (title, series) in enumerate(error_series.items()):
    s = series.dropna()
    for col, plot_fn in enumerate([plot_acf, plot_pacf]):
        ax = axes[row, col]
        kwargs = dict(lags=200, ax=ax, alpha=0.05)
        if col == 1:
            kwargs["method"] = "ywm"
        plot_fn(s, **kwargs)
        ax.set_ylabel(title, fontsize=8)
        for lag in (24, 168):
            ax.axvline(lag, color="red", lw=0.8, linestyle="--")
        ax.set_xlabel("lag")
        if row == 0:
            ax.set_title(["ACF", "PACF"][col], fontsize=12, fontweight="bold")
        else:
            ax.set_title("")

plt.suptitle(
    "ACF and PACF of forecast errors and plan-vs-actual deviations (2024, lags 0–200)\n"
    "Red dashed lines: lag 24h and 168h (smallest actionable lags at gate closure)",
    y=1.01,
)
plt.tight_layout()
plt.show()
```

**Load error:** Strong autocorrelation at short lags, with visible peaks at lag 24h and 168h — systematic under- or over-forecasting persists for one to several days before decaying. **Conclusion: a lagged load forecast error feature is justified.**

**Wind onshore error:** Autocorrelation decays more slowly than load, with a clear peak at lag 24h and a residual signal at lag 168h — weather systems driving forecast errors persist over days. **Conclusion: lagged wind onshore forecast error is useful.**

**Wind offshore error:** Similar persistence to onshore but somewhat lower autocorrelation at lag 168h — offshore wind regimes are driven by larger-scale synoptic patterns. **Conclusion: lagged wind offshore forecast error is likely useful; confirm at feature selection.**

**Solar error (daytime):** Visible autocorrelation at short daytime steps; the 24h/168h reference lines are approximate because lags represent consecutive qualifying steps (solar_forecast > 100 MW), not calendar hours. Persistence reflects multi-day cloud-cover regimes. **Conclusion: a lagged solar forecast error feature is plausible; include conditionally.**

**Cross-border deviation:** Lower but non-negligible autocorrelation at short lags, decaying faster than the weather-driven series. **Conclusion: borderline; include as a candidate and drop if SHAP confirms no contribution.**

**Two critical leakage corrections relative to the EDA:**

**(a) Residual load and renewable share as forecast versions:** The EDA used `load_actual − gen_wind_onshore − gen_wind_offshore − gen_solar` as residual load and the corresponding actuals-based renewable share. At gate closure, these actuals are not yet available for the delivery hours of day D. Feature engineering must derive these quantities from the day-ahead forecasts instead.

**(b) Cross-border flows and forecast errors only lagged:** Scheduled and physical cross-border flows for the delivery hours of day D are either not yet determined (physical) or endogenous to the market coupling process (scheduled). Only lagged values (≥ 24h - 48h) from previous days are leakage-free. The same applies to all forecast errors: the error for hour H on day D is only observed after H has passed.

### 4.4 Open Decisions for Sprint 2

The following decisions must be resolved at the start of Sprint 2, before any feature code is written. Each entry states the question, options, and the owner's preliminary recommendation based on the EDA evidence.

**D1 — Resolution: hourly vs 15-min.**  
*Question:* Should the pipeline operate on hourly or 15-minute intervals?  
*Options:* (a) Resample the post-2025-09-30 15-min data to hourly and keep a single hourly pipeline. (b) Migrate the entire pipeline to 15-min resolution. (c) Leave it as it is and try to work with the resolution break.

*Idea before Feature Engineering:* (a). Even though intra-hour variation is high or it could well be that the market works on both resolutions very similarly, most of the data is hourly. So most of the time, the market calculated hourly prices, therefore that is, what a model can surely learn. It has implications on how to generate the mean hourly price after the resolution break, as it correctly would be the volume weighted average price. A 15-minute-only model could be considered in a second step.

**D2 — Crisis-regime handling.**  
*Question:* How should the 2022–2023 crisis period be treated during training?  
*Options:* (a) Truncate the training window to exclude the crisis. (b) Apply sample weighting to down-weight crisis observations. (c) Include an explicit regime indicator as a feature.  
*Idea before Feature Engineering:* (c) Start with a dummy and drop it, if it is not useful.

**D3 — Block vs recursive forecast and the 1h lag.**  
*Question:* Should the model produce a 24-hour block forecast or a recursive 1-step-ahead forecast?  
*Options:* (a) Block forecast (one model, 24 target hours simultaneously or separate per hour). (b) Recursive: forecast hour by hour, feeding predicted prices as inputs.  
*Idea before Feature Engineering:* (a) Block forecast, as recursive prediction errors would compound. Consequence: `price_lag_1h` is leakage for a block model targeting all 24 hours of day D — the previous hour on day D is not available at gate closure. Use lags ≥ 24h only. This also simplifies the backtesting walk-forward setup (one prediction per day, not 24 sequential steps).

**D4 — EUA CO2 missingness before Oct-2021 (~38% of training data).**  
*Question:* How to handle the ~38% NaN period for EUA CO2?  
*Options:* (a) Truncate training to ≥ Oct-2021 (loses the “calm” regime). (b) Keep full range with a `eua_missing` indicator flag. (c) Source EUA data from EEX/ICE for the pre-2021 period.  
*Idea before Feature Engineering:* Start with option (b): include a `eua_missing` binary flag so the model can learn a different price equation for the pre-2021 window. If the model fails to learn a useful representation of the calm regime, switch to (a). Option (c) is the production-grade solution but out of scope for the portfolio project.

**D5 — Target transformation.**  
*Question:* Should the day-ahead price be transformed before modelling?  
*Options:* (a) No transformation (raw EUR/MWh). (b) `asinh` (defined for negative values and zero, log-like for large positives). (c) Signed log: `sign(x) * log(1 + |x|)`. (d) Quantile/rank transform.  
*Idea before Feature Engineering:* No transformation for LightGBM (tree models are scale-invariant and handle skew natively). For an ARIMAX or linear baseline, consider `asinh` given the negative prices. For quantile regression (Pinball loss), no transformation is needed. Decide per model type in Sprint 2/3.

**D6 — Forecast-error features: Who to include and how.**  
*Question:* Which lagged forecast-error features should be included? Is it smarter to enter them directly as model inputs or use them to build a pre-model to best forecast residual load?  
*Evidence from 4.2:* Load error — clear persistence at lag 24h/168h → **include**. Wind onshore error — clear persistence → **include**. Wind offshore error — moderate persistence → **include, verify with SHAP**. Solar error — plausible persistence (daytime lags) → **include conditionally, verify with SHAP**. Cross-border deviation — weaker signal → **include, drop if SHAP confirms no contribution**.  Pre-Model? → Test if forecasts can be improved in Feature Engineering.
*Idea before Feature Engineering:* Include all five as candidates in the initial feature set; prune via SHAP importance in Sprint 3; Test a residual load forecast model.

### 4.5 Summary and Hand-off to Feature Engineering

**Data quality (section 2):** The DE-LU dataset is structurally sound. The main data-quality constraints that feature engineering must handle are: the mixed hourly/15-min resolution break at 2025-09-30 (→ D1), EUA CO2 missingness before Oct-2021 (→ D4), and the permanent disappearance of `gen_nuclear` after April 2023 (treat as a known structural zero). Negative prices are systematic, not noise — driven by high renewable share at low demand, growing year-on-year. The 2022 crisis constitutes a distinct regime in both level and volatility; the gas price feature should allow a tree model to capture this without a hard regime dummy.
